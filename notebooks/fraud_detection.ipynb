```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
np.random.seed(42)

n = 20000

df = pd.DataFrame({
    "amount": np.random.exponential(scale=100, size=n),
    "hour": np.random.randint(0, 24, size=n),
    "distance_from_home": np.random.exponential(scale=20, size=n),
    "failed_attempts": np.random.poisson(lam=0.3, size=n),
    "country_changed": np.random.binomial(1, 0.05, size=n)
})
```

```python
df.head()
```

```python
df.shape
```

```python
fraud_probability = (
    0.01
    + 0.0005 * df["amount"]
    + 0.01 * df["distance_from_home"]
    + 0.15 * df["failed_attempts"]
    + 0.8 * df["country_changed"]
    + 0.3 * ((df["hour"] < 5).astype(int))
)

fraud_probability = np.clip(fraud_probability, 0, 0.95)

df["is_fraud"] = np.random.binomial(1, fraud_probability)
```

```python
df["fraud_probability"] = fraud_probability
```
